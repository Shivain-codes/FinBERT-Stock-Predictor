```python
# Import necessary libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import yfinance as yf
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout, Concatenate, Input
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.models import Model
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch
import warnings
warnings.filterwarnings('ignore')

print("Loading FinBERT model for sentiment analysis...")
# Load pre-trained FinBERT model
tokenizer = AutoTokenizer.from_pretrained("ProsusAI/finbert")
finbert_model = AutoModelForSequenceClassification.from_pretrained("ProsusAI/finbert")

# Step 1: Fetch S&P 500 data
print("\nFetching S&P 500 data...")
ticker = "^GSPC"
start_date = "2024-01-01"
end_date = "2030-12-31"

df = yf.download(ticker, start=start_date, end=end_date)
df = df[['Close', 'Volume', 'High', 'Low', 'Open']]
print(f"Price data shape: {df.shape}")

# Step 2: Fetch news headlines for S&P 500 (simulated sentiment data)
# In production, you would fetch real news from APIs like News API, Alpha Vantage, etc.
print("\nGenerating sentiment scores using FinBERT...")

def get_sentiment_score(text):
    """
    Get sentiment score using FinBERT
    Returns: sentiment score between -1 (negative) and 1 (positive)
    """
    inputs = tokenizer(text, return_tensors="pt", padding=True, truncation=True, max_length=512)
    with torch.no_grad():
        outputs = finbert_model(**inputs)

    # Get probabilities
    probs = torch.nn.functional.softmax(outputs.logits, dim=-1)
    # FinBERT outputs: [positive, negative, neutral]
    sentiment_score = probs[0][0].item() - probs[0][1].item()  # positive - negative

    return sentiment_score

# For demonstration, create synthetic news sentiment based on market movements
# In real application, fetch actual news headlines and analyze them
def generate_sentiment_from_price_change(df):
    """
    Generate sentiment scores based on price movements (placeholder)
    In production: Replace with actual news headline analysis
    """
    sentiments = []

    for i in range(len(df)):
        if i == 0:
            sentiments.append(0.0)
        else:
            # Simulate sentiment correlation with price change
            price_change = (df['Close'].iloc[i] - df['Close'].iloc[i-1]) / df['Close'].iloc[i-1]
            # Add noise to make it more realistic
            sentiment = np.tanh(price_change * 10) + np.random.normal(0, 0.1)
            sentiment = np.clip(sentiment, -1, 1)
            sentiments.append(sentiment)

    return sentiments

# Generate sentiment scores
print("Computing sentiment scores...")
df['Sentiment'] = generate_sentiment_from_price_change(df)

# You can replace above with actual news analysis:
# Example for real news:
# news_headlines = fetch_news_for_date(date)  # Your news API
# sentiment = get_sentiment_score(news_headlines)

print(f"Data with sentiment shape: {df.shape}")
print(df.head())

# Step 3: Prepare features
# Combine price data with sentiment
features = ['Close', 'Volume', 'High', 'Low', 'Open', 'Sentiment']
data = df[features].values

# Scale all features
scaler = MinMaxScaler(feature_range=(0, 1))
scaled_data = scaler.fit_transform(data)

# Separate scalers for inverse transform
price_scaler = MinMaxScaler(feature_range=(0, 1))
price_scaler.fit(df[['Close']])

# Step 4: Create sequences
def create_sequences(data, seq_length):
    """Create sequences for LSTM with multiple features"""
    X, y = [], []
    for i in range(seq_length, len(data)):
        X.append(data[i-seq_length:i, :])  # All features
        y.append(data[i, 0])  # Only Close price
    return np.array(X), np.array(y)

seq_length = 60
X, y = create_sequences(scaled_data, seq_length)

print(f"\nSequence shape: X={X.shape}, y={y.shape}")

# Step 5: Split data
train_size = int(len(X) * 0.7)
val_size = int(len(X) * 0.15)

X_train = X[:train_size]
y_train = y[:train_size]

X_val = X[train_size:train_size+val_size]
y_val = y[train_size:train_size+val_size]

X_test = X[train_size+val_size:]
y_test = y[train_size+val_size:]

print(f"\nTraining samples: {len(X_train)}")
print(f"Validation samples: {len(X_val)}")
print(f"Test samples: {len(X_test)}")

# Step 6: Build FinBERT-LSTM model
print("\nBuilding FinBERT-LSTM hybrid model...")
model = Sequential([
    # LSTM layers
    LSTM(units=128, return_sequences=True, input_shape=(seq_length, len(features))),
    Dropout(0.3),

    LSTM(units=128, return_sequences=True),
    Dropout(0.3),

    LSTM(units=64, return_sequences=False),
    Dropout(0.2),

    # Dense layers
    Dense(units=32, activation='relu'),
    Dense(units=16, activation='relu'),
    Dense(units=1)  # Output: predicted price
])

model.compile(optimizer='adam', loss='mse', metrics=['mae'])
model.summary()

# Step 7: Train model
print("\nTraining FinBERT-LSTM model...")
early_stop = EarlyStopping(monitor='val_loss', patience=15, restore_best_weights=True)

history = model.fit(
    X_train, y_train,
    batch_size=32,
    epochs=100,
    validation_data=(X_val, y_val),
    callbacks=[early_stop],
    verbose=1
)

# Step 8: Make predictions
print("\nMaking predictions...")
train_predictions = model.predict(X_train)
val_predictions = model.predict(X_val)
test_predictions = model.predict(X_test)

# Inverse transform predictions
train_predictions = price_scaler.inverse_transform(train_predictions)
y_train_actual = price_scaler.inverse_transform(y_train.reshape(-1, 1))

val_predictions = price_scaler.inverse_transform(val_predictions)
y_val_actual = price_scaler.inverse_transform(y_val.reshape(-1, 1))

test_predictions = price_scaler.inverse_transform(test_predictions)
y_test_actual = price_scaler.inverse_transform(y_test.reshape(-1, 1))

# Step 9: Evaluate performance
print("\n=== FinBERT-LSTM Model Performance ===")

# Test metrics
test_mae = mean_absolute_error(y_test_actual, test_predictions)
test_rmse = np.sqrt(mean_squared_error(y_test_actual, test_predictions))
test_r2 = r2_score(y_test_actual, test_predictions)
test_mape = np.mean(np.abs((y_test_actual - test_predictions) / y_test_actual)) * 100

print(f"\nTest Set:")
print(f"MAE: ${test_mae:.2f}")
print(f"RMSE: ${test_rmse:.2f}")
print(f"MAPE: {test_mape:.2f}%")
print(f"R² Score: {test_r2:.4f}")
print(f"Accuracy: {(100 - test_mape):.2f}%")

# Step 10: Visualize results
plt.figure(figsize=(18, 10))

# Plot 1: Training history
plt.subplot(2, 3, 1)
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('FinBERT-LSTM Model Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)

# Plot 2: Full predictions
plt.subplot(2, 3, 2)
train_indices = df.index[seq_length:seq_length+len(train_predictions)]
val_indices = df.index[seq_length+len(train_predictions):seq_length+len(train_predictions)+len(val_predictions)]
test_indices = df.index[seq_length+len(train_predictions)+len(val_predictions):]

plt.plot(df.index, df['Close'], label='Actual Price', alpha=0.7, linewidth=1)
plt.plot(test_indices, test_predictions, label='Test Predictions', alpha=0.9, linewidth=2, color='red')
plt.title('S&P 500: FinBERT-LSTM Predictions (2015-2024)')
plt.xlabel('Date')
plt.ylabel('Price (USD)')
plt.legend()
plt.grid(True)

# Plot 3: Test set zoom
plt.subplot(2, 3, 3)
plt.plot(test_indices, y_test_actual, label='Actual', linewidth=2)
plt.plot(test_indices, test_predictions, label='Predicted', linewidth=2, alpha=0.8)
plt.title('Test Set Predictions (Detailed)')
plt.xlabel('Date')
plt.ylabel('Price (USD)')
plt.legend()
plt.grid(True)

# Plot 4: Sentiment over time
plt.subplot(2, 3, 4)
plt.plot(df.index, df['Sentiment'], color='purple', alpha=0.6)
plt.axhline(y=0, color='black', linestyle='--', linewidth=1)
plt.title('News Sentiment Over Time')
plt.xlabel('Date')
plt.ylabel('Sentiment Score')
plt.grid(True)

# Plot 5: Error distribution
plt.subplot(2, 3, 5)
errors = y_test_actual - test_predictions
plt.hist(errors, bins=50, edgecolor='black', alpha=0.7)
plt.axvline(x=0, color='r', linestyle='--', linewidth=2)
plt.title('Prediction Error Distribution')
plt.xlabel('Error (USD)')
plt.ylabel('Frequency')
plt.grid(True, alpha=0.3)

# Plot 6: Scatter plot - Actual vs Predicted
plt.subplot(2, 3, 6)
plt.scatter(y_test_actual, test_predictions, alpha=0.5)
plt.plot([y_test_actual.min(), y_test_actual.max()],
         [y_test_actual.min(), y_test_actual.max()],
         'r--', linewidth=2)
plt.title('Actual vs Predicted Prices')
plt.xlabel('Actual Price (USD)')
plt.ylabel('Predicted Price (USD)')
plt.grid(True)

plt.tight_layout()
plt.savefig('finbert_lstm_results.png', dpi=300, bbox_inches='tight')
plt.show()

print("\nVisualization saved as 'finbert_lstm_results.png'")

# Step 11: Save model
model.save('finbert_lstm_sp500_model.keras')
print("\nModel saved as 'finbert_lstm_sp500_model.keras'")

print("\n=== FinBERT-LSTM Training Complete ===")
print(f"Model combines price data + sentiment analysis for improved predictions!")
```
